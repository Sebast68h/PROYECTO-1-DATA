# Bienestar y caracteristicas de la poblacion 


## pregunta analitica

¿Cómo varía el bienestar de las personas de 18 años o más según su sexo y su estado civil, y qué relación tiene con su estado emocional del día anterior?

# DIC variables
#### P6020 = Sexo al nacer  (NOMINAL)
#### P6040 = Años Cumplidos (NUMERICA-discreta)
#### P1927 = Escalon de vida de 0-10 (ordinal)
#### P1899 = Satisfaciion con el trabajo actual 0-10 (ordinal)
#### P3175 = Satisfaccion con el tiempo libre 0-10(ordinal)
#### P1898 = Satisfaccion con la seguridad 0-10(ordinal)
#### P1897 = Satisfaccion con la salud 0-10(ordinal)
#### P1896 = Satisfaccion con el ingreso 0-10(ordinal)
#### P1895 = Satisfaciion con su vida en general 0-10 (ordinal)
#### P1905 = Consideracion de que las cosas que en su vida valen la pena 0-10 (ordinal)
#### P6074 = siempre ha vivido en el municipio(NOMINAL)
#### P5502 = Estado civil (NOMINAL)
#### P1901 = Qué tan feliz se sintió el día de ayer 0-10 (ordinal)
#### P1903 = Qué tan preocupado/a se sintió el día de ayer 0-10 (ordinal)
#### P1904 = Qué tan triste se sintió el día de ayer 0-10 (ordinal)
#### bienestar = Promedio de P1895, P1896, P1897, P1898, P1899 y P3175 (variable construida, 0-10)
#### tristeza = promedio de P1903 y P1904 (preocupado y triste), 0-10 (más alto = más tristeza)
#### felicidad = igual a P1901, qué tan feliz se sintió el día de ayer 0-10 (más alto = más feliz)


Nota: las variables P1895, P1896, P1897, P1898, P1899 y P3175 preguntan qué tan satisfecha está la persona con distintas partes de su vida (vida en general, ingreso, salud, seguridad, trabajo y tiempo libre). Con esas seis armamos el índice de bienestar.

Las variables P1901, P1903 y P1904 preguntan cómo se sintió la persona el día de ayer (feliz, preocupada, triste). Esas no entran en el índice, las usamos aparte para compararlas con el bienestar.

P1927 y P1905 también son del módulo de bienestar, pero no las usamos ni en el índice ni en la comparación.

```python
var = ["P6020","P6040","P1927","P1899","P3175","P1898","P1897","P1896","P1895","P1901","P1903","P1904","P5502","P1905"]
usable = datos_limpios[var].copy()
usable
```

```python
usable["P1896"] = usable["P1896"].replace(99, pd.NA)
usable["P1896"] = pd.to_numeric(usable["P1896"], errors="coerce")
```

```python
variables = ["P1927","P1899","P3175","P1898","P1897","P1896", "P1901", "P1903","P1895","P1904", "P1905"]
datos_limpios[variables].corr()
```

```python
variables = ["P1927","P1899","P3175","P1898","P1897","P1896", "P1901", "P1895", "P1903", "P1904", "P1905"]
datos_limpios[variables].corr(method= "spearman")
```

```python
variables_bienestar = ["P1895", "P1896", "P1897", "P1898", "P1899", "P3175"]
usable["bienestar"] = usable[variables_bienestar].mean(axis=1)
usable[["bienestar"] + variables_bienestar]

variables_emocionales = ["P1903", "P1904"]
usable["tristeza"] = usable[variables_emocionales].mean(axis=1)

variables_emocionales = ["P1901"]
usable["felicidad"] = usable[variables_emocionales].mean(axis=1)
```

```python
usable["bienestar"] = pd.to_numeric(usable["bienestar"], errors="coerce")
usable["tristeza"] = pd.to_numeric(usable["tristeza"], errors="coerce")
usable["felicidad"] = pd.to_numeric(usable["felicidad"], errors="coerce")
usable["P1896"].describe()
```

```python
nombre_archivo = "usable.csv"
carpeta_destino = os.path.join("..", "Data", "procesed")
os.makedirs(carpeta_destino, exist_ok=True)
ruta_salida = os.path.join(carpeta_destino, nombre_archivo)
usable.to_csv(ruta_salida, index=False, encoding="utf-8-sig")
```

```python
def guardar_img(nombre):
    """Guarda el gráfico en la carpeta y después lo muestra."""
    plt.tight_layout()
    plt.savefig("../outputs/figures/" + nombre)
    plt.show()
```

```python
estado_civil["bienestar_promedio"].plot(kind="bar", title="Bienestar por estado civil", ylim=(7, 9), xlabel="Estado civil")
plt.ylabel("Bienestar")
guardar_img("estado_civil_barras.png")
```

```python
estado_civil["tristeza_promedio"].plot(kind="bar", title="Tristeza por estado civil", ylim=(2, 3), xlabel="Estado civil")
plt.ylabel("Tristeza")
guardar_img("estado_civil_tristeza_barras.png")
```

```python
estado_civil["felicidad_promedio"].plot(kind="bar", title="Felicidad por estado civil", ylim=(7, 9), xlabel="Estado civil")
plt.ylabel("Felicidad")
guardar_img("estado_civil_felicidad_barras.png")
```

```python
usable["bienestar"].plot(kind="hist", bins=20, title="Distribución del bienestar")
plt.xlabel("Bienestar (0-10)")
guardar_img("bienestar_histograma.png")
```
